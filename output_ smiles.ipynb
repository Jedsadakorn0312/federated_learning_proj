```python
import numpy as np
from rdkit import Chem
from rdkit.Chem import rdFingerprintGenerator

FP_SIZE = 1024
mfgen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=FP_SIZE)

def smiles_to_vector(smiles: str, fp_size=FP_SIZE) -> np.ndarray:
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return np.zeros(fp_size, dtype=np.float32)
    fp = mfgen.GetFingerprintAsNumPy(mol)   # numpy array 0/1
    return fp.astype(np.float32)

# ตัวอย่าง
smiles = "CC(=O)Oc1ccccc1C(=O)O"  # Aspirin
vec = smiles_to_vector(smiles)
print("Vector shape:", vec.shape)
print("First 20 dims:", vec[:20])
```
